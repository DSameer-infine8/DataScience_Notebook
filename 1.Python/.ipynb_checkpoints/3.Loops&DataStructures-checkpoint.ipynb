## For Loop

- Q1. Accept a  integer and print hello world n times.

```python
n = int(input("Enter a number:"))

for i in range(1,n): 
    print("Hello World!")
```

- Q2. Print natural numbers up to n

```python
n = int(input("Enter a number:"))

for i in range(1,n+1): 
    print(i)
```

- Q3. Reverse print n to 1.

```python
n = int(input("Enter a number:"))

for i in range(n,0,-1): 
    print(i)
```

- Q4. Take a number & print its table 

```python
num = int(input("Enter the number for Tables:"))

for i in range(1,11): 
    print(f'{num} x {i} = {num*i}')
```

- Q5. Sum up to n terms

```python
n = int(input("Enter the number:"))
sum = 0
for i in range(1,n+1): 
    sum +=i

print(sum)
```

- Q6. Factorial of a number

```python
n = int(input("Enter the number:"))
fact = 1
for i in range(1,n+1): 
    fact *=i 

print(fact)
```

- Q7. Print the sum of all even and odd numbers in a range seprately 

```python
n = int(input("Enter the number:"))
even = 0
odd = 0
for i in range(1,n+1): 
    if i%2==0: 
        even +=i
    else:
        odd +=i

print(f'Even sum={even} & odd sum={odd}')
```

- Q8. Print all the facotrs of a number

- Q9. Accept a number and check if it is a perfect number or not

```python
n=int(input("Enter the number to check perfect number:"))

sum = 0
for i in range(1,n): 
    if n%i==0: 
        print(i)
        sum+=i

if(sum==n): 
    print('The given number is perfect number')
else:
    print('It is not perfect number')
```

- Q10. Check weather the number is prime or not

```python
# Prime is a number which only divides by 1 and itself only

n = int(input("Check for prime no:"))

count = 0
for i in range(1,n+1): 
    if n%i == 0: 
        count +=1

if count ==2: 
    print("Your number is prime")
else:
    print("Your number is not prime")
```

- Q11. Reverse a string without using in-build functions

```python
a = 'SAMEER'
print(a[::-1])

b=''
for i in range(len(a)-1,-1,-1): 
    print(a[i])
    b += a[i]

print(b)
```

- Q13. Count all the letters,digits and special symbols from a give string.

```python
a = 'sdjbf139uff9@*$9gi*!@'

char =0
dig =0
splChar =0
for i in a: 
    if i.isdigit():
        dig +=1
    elif i.isalpha():
        char+=1
    else:
        splChar +=1

print(f"Your digits are {dig}\n Your alphabets are {char}\n Your special characters are {splChar}")
```

## Note: When to use For loop and while loop?
### Use a for loop when you know exactly how many times you want to repeat something.

- Simple examples:
  - Loop through a list of items (fruits, numbers, names).
  - Repeat something 10 times, 5 times, etc.
  - Go through each row in a dataset or each character in a string.

### Use a while loop when you don’t know how many times the loop should run and you want it to continue until a condition becomes false.

- Simple examples:
  - Keep asking the user for input until they type “exit”.
  - Keep looping until a sensor reaches a value.
  - Keep running until some condition in your program changes.

## While Loop

- Q1. Print 1 to 30 numbers

```python
a = 1

while a<=30: 
    print(a)
    a+=1
```

- Q2. Separate each digit of number and print it on the new line

```python
a = 2359

while a>0: 
    print(a%10)
    a //=10
```

- Q3. Accept a number and print its reverse 

```python
a = int(input("Enter a number:"))

rev= 0
while a>0:
    rev = rev*10+a%10
    a=a//10

print(rev)
```

- Q4. Accept a number and check if it is a pallindrome or not

```python
a = int(input("Enter a number:"))
copy = a

rev= 0
while a>0:
    rev = rev*10+a%10
    a=a//10

if copy==rev:
    print("It is a Pallindrome")
else:
    print("It is not Pallindrome")
```

- Q5. Create a random number guessing game with python

```python
import random

num = random.randint(1,10)
tries = 0

while True:
    guess = int(input("Guess your number:"))
    if num == guess:
        tries+=1
        print("Your are right ")
        break
    elif num < guess:
        print("Go a little lower")
        tries +=1
    elif num > guess:
        print("Go a little higher")
    else:
        tries+=1
        print("Sorry you are wrong")
```

### List -> Mutable, Duplicates, Ordered, Hetrogenous.

- Q1. Print +ve & -ve elements of an list.

```python
l = [-45,67,12,-68,-69,34]

print("positive elements are ")
for i in l:
    if i >= 0:
        print(i)
print("negitive elements are")

for i in l:
    if i < 0:
        print(i)
```

- Q2. Mean of list elements.

```python
l = [12,435,67,89,23,25,69]

sum = 0

for i in l:
    sum = sum + i

print(sum/len(l))
```

- Q3. Find the greatest element & print its index too.

```python
l = [12,43,67,89,23,25,69,11,456,56]
great = 0
for i in range(0,len(l)):
    if l[i] > great:
        great = l[i]

print(great)
```

- Q4. Find the second greatest element.

```python
l = [92,12,43,67,89,23,25,69,11,456,56]
great = 0
sec = 0
for i in range(0,len(l)):
    if l[i] > great:
        sec = great
        great = l[i]

print(sec)
print(great)
```

- Q5. Check if the list is sorted.

```python
a = [12,13,18,15,16]

for i in range(len(a)-1):
    if a[i] < a[i+1]:
        continue
    else:
        print("your list is not sorted")
        break
else:
    print("your list is sorted")
```

#### Only difference between list and tuples are that Tuples values cant be changed so it is called immutable.

- Only 2 methods of tuples

```python
t = (5,3,6,32,4,6,4,5,9)

print(t.index(4))
print(t.count(5))
```

```python
a = (1,2,3,4,5,5,5.5,print(),"hello")

count = a.count(5)
print(count)

a = (1,)
print(type(a))
```

### Set -> Mutable, No Duplicates, Unordered, Semi-Hetrogenous.

- Note: As sets are unordered , can't be used for loops to print individual value, if you do it will print random values

```python
a = {1,8,9,"hello",2,3,4,5}

for i in a:
    print(i)
```

```python
a = {8,1,2,3,4}
a.clear()
print(a)
```

- Note: Set used for statistics

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}

#---------------------
print(a.union(b))   
print(a|b)

print(a.intersection(b))
print(a&b)

print(a.difference(b))
print(a-b)

print(b.difference(a))
print(b-a)

print(a.symmetric_difference(b))
print(a^b)
```

### Dictionary -> Mutable, Duplicates, Ordered, Hetrogenous.

- Note: Can perform CRUD operations on Dict only on values and not keys.

```python
d = {10:100,20:200,30:300,40:400}

d[10] = 99 #updating
d[50] = 500 # creating
del d[30] # deleting 

print(d)
```

```python
d = {10:100,20:200,30:300,40:400}

print(d.items())
```

```python
# Access keys and values

d = {10:100,20:200,30:300,40:400}

for i in d:
    print(f'key:{i}')   #Prints all keys
    print(f'Value:{d[i]}')   #Prints all Values

#loop on values
for i in d.values(): 
    print(i)
```

```python
# Deep Copy

a=[1,2,3,4,5]
b=a
b[0]=100
print(a)
```

```python
# Shallow Copy
a=[1,2,3,4,5]
b=a.copy()
b[0]=100
print(a)
```

```python
d1 = {10:100,20:200,40:300}
d2 = {40:400,50:500,60:600}


for i in d2:
    d1[i] = d2[i]

print(d1)
```

```python
d1 = {10:100,20:200,40:300}
sum = 0

for i in d1:
    sum = sum + d1[i]

print(sum)
```

```python
a = [1,1,1,2,2,2,3,3,3,4,4,4,5,5,6,7,8]

d = {}
for i in a:
    if i in d.keys():
        d[i] +=1 
    else:
        d[i] = 1

print(d)
```
