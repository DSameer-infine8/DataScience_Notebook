## plotly
- it is a Python library which is used to design graphs, especially interactive graphs. It can plot various graphs and charts like histogram, barplot, boxplot, spreadplot and many more. It is mainly used in data analysis as well as financial analysis. plotly is an interactive visualization library.y.

## cufflinks
- connects plotly with pandas to create graphs and charts of dataframes directly.

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
import cufflinks as cf 
from plotly.offline import iplot
```

```python
data = sns.load_dataset('tips')
```

```python
data['total_bill'].iplot()
```

```python
import cufflinks as cf
import plotly.offline as pyo

pyo.init_notebook_mode(connected=True)
cf.go_offline()
```

```python
data['total_bill'].iplot()
```

```python
data.groupby('day')['tip'].mean().iplot(kind='bar')
```

```python
data.iplot(kind='scatter',x='total_bill', y='tip',mode = 'markers')
```

```python
#data.iplot(kind='pie',x='total_bill', y='tip', labels='sex')
data[['total_bill','day']].iplot(kind = 'box',x="total_bill",y='day')
```
