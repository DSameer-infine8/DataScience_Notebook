```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

## Heat map

```python
flights = sns.load_dataset('flights')
```

```python
flights
```

```python
tips = sns.load_dataset("tips")
```

```python
tips
```

```python
tips_corr = tips[['total_bill','tip','size']]
```

```python
tips_corr.corr()
```

```python
sns.heatmap(tips_corr.corr(),annot= True)
```

## Cluster Map

```python
sns.clustermap(tips_corr.corr(),annot= True)
```

## Pivot  table Heat Map

```python
pvtflight = flights.pivot_table(values="passengers",index= 'month', columns="year")
```

```python
pvtflight
```

```python
sns.heatmap(pvtflight)
```

## Regression Plot

### Implot() 

Regression plot is used in ML so we see how it is created and what are the values we can put inside it

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

```python
sns.lmplot(x= 'total_bill',y='tip',data=tips)
```

```python
sns.lmplot(x= 'total_bill',y='tip',data=tips,hue='sex')
```

```python
sns.lmplot(x= 'total_bill',y='tip',data=tips,hue='sex',markers=['o','v'])
```

```python
sns.lmplot(x= 'total_bill',y='tip',data=tips,hue='sex',markers=['o','v'],palette='cool')
```

```python
sns.lmplot(x= 'total_bill',y='tip',data=tips,hue='sex',markers=['o','v'],scatter_kws={'s':100})
```
