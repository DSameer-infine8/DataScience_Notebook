## Distribution Plots
##### 
We use distribution plots to visualize the distribution of quantitative data##### 

Let's discuss some plots that allow us to visualize the distribution of a data set. These plotre:s- distplot/histplot
- jointplot
- pairplot and
- rugplotugplot
  rugplot



```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()    #It is also a Qualitative has cant perform distribution plots
```

**Note:**
- Hence sex,smoker,day,time and size are all Qualitative value. Distribution plots cant be performed.
- only on total_bill and tip  distribution plots can be performed

```python
sns.displot(df['total_bill'])
```

```python
sns.displot(df['tip'])
```

```python
sns.displot(df['total_bill'],bins=1)
```

```python
sns.displot(df['total_bill'],bins=4)
```

```python
sns.displot(df['total_bill'],bins=20)
```

```python
sns.displot(df['total_bill'],bins=100)
```

- Note: histplot = displot (both are same)

```python
sns.histplot(df['total_bill'],bins=20)
```

```python
sns.histplot(df['total_bill'],bins=20,kde= True)  #default value bins=20
```

- Note: Prefer to use histplot over displot

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20,kde= True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20,kde= True)
```

```python
sns.jointplot(x='total_bill',y='tip', data=df,kind = "scatter")
```

```python
sns.jointplot(data=df, x="tip", y="tip")
```

```python
sns.jointplot(x='total_bill',y='tip', data=df,kind = 'reg')
```

```python
sns.jointplot(x='total_bill',y='tip', data=df,kind = '')
```

```python
#Important 

sns.pairplot(df) 
```

- Only numerical values(Quantitative) are selected and ploted

```python
sns.pairplot(df,hue="sex")
```

```python
sns.pairplot(df,hue="size",palette="rainbow")
```

```python
# Not so important 
sns.rugplot(df)
```

## Catagorical Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

    boxplot
    violinplot
    stripplot
    swarmplot
    barplot
    countplot

```python
sns.countplot(x =df['sex'])
```

```python
sns.countplot(x =df['sex'],hue=df["smoker"])
```

```python
sns.barplot(x = df['sex'],y= df["tip"])
```

#### barplot is unique here we need to give to values for x and y axis
### - In x axis the values should be "Qualitative" and 
### -  y axis the values should be "Quantitative"
#### Ex: sns.barplot(x = df['sex'],y= df["tip"])

```python
sns.barplot(x = df['sex'],y= df['total_bill'])
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'],estimator=np.sum)
```

```python
sns.barplot(x=df['sex'],y=df['tip'],estimator=np.sum)
```

```python
sns.boxplot(x='tip',y='day',data=df)
```


boxplot and violinplot

boxplots and violinplots are used to shown the distribution of categorical data. A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be “outliers” using a method that is a function of the inter-quartile range.


```python
sns.violinplot(x='day',y='total_bill',data=df)
```


stripplot and swarmplot

The stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.

The swarmplot is similar to stripplot(), but the points are adjusted (only along the categorical axis) so that they don’t overlap. This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations (both in terms of the ability to show all the points and in terms of the computation needed to arrange them).


```python
sns.stripplot(x = 'day',y = 'total_bill',data = df)
```

```python
sns.swarmplot(x = 'day',y = 'total_bill',data = df)
```

```python
sns.violinplot(x='tip',y= 'day',data=df)
sns.swarmplot(x='tip',y= 'day',data=df,color='purple')
```
