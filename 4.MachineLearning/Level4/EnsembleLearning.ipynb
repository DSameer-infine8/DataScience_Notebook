```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 

import warnings  
warnings.filterwarnings('ignore')
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression 
from sklearn.tree import DecisionTreeClassifier 
from sklearn.svm import SVC 
from sklearn.model_selection import train_test_split 
from sklearn.metrics import accuracy_score, classification_report
from sklearn.ensemble import StackingClassifier 
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
X = df.drop('species',axis=1)
y = df['species']
```

```python
le = LabelEncoder() 
y_encoded = le.fit_transform(y)
```

```python
y_encoded.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42, stratify= y_encoded)
```

## Stacking

```python
base_learners = [
    ('dt', DecisionTreeClassifier(random_state=42)), 
    ('svc', SVC(probability=True, kernel='rbf', random_state=42)), 
    ('lr', LogisticRegression(max_iter=1000))
]
```

```python
meta_learner = LogisticRegression(max_iter=1000) 
```

```python
stacking_clf = StackingClassifier(
    estimators= base_learners, 
    final_estimator= meta_learner,cv=5
)
```

```python
stacking_clf.fit(X_train, y_train)
```

```python
y_pred = stacking_clf.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

## Bagging

- In Bagging you are taking One same ML algo and training many model of different random data each time.

### Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,    #number of tree 
    max_depth=None,      # Let trees grow fully, no depth restriction
    random_state=42
)
```

```python
rf_model.fit(X_train, y_train)
```

```python
y_rf_pred = rf_model.predict(X_test)
```

```python
accuracy_score(y_test, y_rf_pred)
```

```python
print(classification_report(y_test, y_rf_pred))
```

## Boosting

- A boosting is a technique in which we will use concurrent data (same data again and again) and providing errors of previous model to next model until weak learner models turns to strong learners

```python
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
```

```python
from xgboost import XGBClassifier
```

```python
ada_model = AdaBoostClassifier(n_estimators=100, random_state=42)
```

```python
ada_model.fit(X_train, y_train)
```

```python
y_ada_pred = ada_model.predict(X_test)
```

```python
accuracy_score(y_test, y_ada_pred)
```

```python
print(classification_report(y_test, y_ada_pred))
```

```python
gb_model = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1,random_state=42)
```

```python
gb_model.fit(X_train, y_train)
```

```python
y_gb_pred = gb_model.predict(X_test)
```

```python
accuracy_score(y_test, y_gb_pred)
```

```python
print(classification_report(y_test, y_gb_pred))
```

```python
xgb_model = XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth = 3,use_label_encoder=False,
                        eval_metric='mlogloss', random_state=42)
```

```python
xgb_model.fit(X_train, y_train)
```

```python
y_xgb_pred = xgb_model.predict(X_test)
```

```python
accuracy_score(y_test, y_xgb_pred)
```

```python
print(classification_report(y_test, y_xgb_pred))
```
