## Text Preprocessing/Cleaning  for ML

### Step1: Lower Case

```python
# To make short function lambda is used 

df["title"] = df["title"].apply(lambda x : x.lower())
```

```python
df.head()
```

## Step2: Remove Punctuation

```python
import string 

def remove_punchu(txt): 
    return txt.translate(str.maketrans("","",string.punctuation))
```

```python
df["title"] = df['title'].apply(remove_punchu)
```

```python
df.head()
```

## Step3: Removing Numbers 

```python
def remove_nums(txt): 
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new 

df['title'] = df['title'].apply(remove_nums)
```

```python
df.head()
```

## Removing Emojis and Special Characters 

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new +=i
    return new 

df['title'] = df['title'].apply(remove_emojis)
```

```python
df.head()
```

## Remove Stopwords

- Note: Doing all the other tasks are easy but for removing the stop words like 'is','the','are' etc we have to use a library and that is NLTK(Natural Language Toolkit).

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download("punkt")
nltk.download("stopwords")
```

```python
stop_words = set(stopwords.words('english'))
stop_words
```

```python
len(stop_words)
```

```python
df
```

```python
def remove(txt): 
    words = txt.split()
    cleaned = []
    for i in words:
        if not i in stop_words: 
            cleaned.append(i)
    return " ".join(cleaned)
```

```python
df['title'] = df['title'].apply(remove)
```
