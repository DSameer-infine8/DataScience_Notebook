## Text Preprocessing/Cleaning  for ML

### Step1: Lower Case

```python
# To make short function lambda is used 

df["title"] = df["title"].apply(lambda x : x.lower())
```

```python
df.head()
```

## Step2: Remove Punctuation

```python
import string 

def remove_punchu(txt): 
    return txt.translate(str.maketrans("","",string.punctuation))
```

```python
df["title"] = df['title'].apply(remove_punchu)
```

```python
df.head()
```

## Step3: Removing Numbers 

```python
def remove_nums(txt): 
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new 

df['title'] = df['title'].apply(remove_nums)
```

```python
df.head()
```

## Removing Emojis and Special Characters 

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new +=i
    return new 

df['title'] = df['title'].apply(remove_emojis)
```

```python
df.head()
```

## Remove Stopwords

- Note: Doing all the other tasks are easy but for removing the stop words like 'is','the','are' etc we have to use a library and that is NLTK(Natural Language Toolkit).

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download("punkt")
nltk.download("stopwords")
```

```python
stop_words = set(stopwords.words('english'))
stop_words
```

```python
len(stop_words)
```

```python
df
```

```python
def remove(txt): 
    words = txt.split()
    cleaned = []
    for i in words:
        if not i in stop_words: 
            cleaned.append(i)
    return " ".join(cleaned)
```

```python
df['title'] = df['title'].apply(remove)
```

```python
df.head()
```

## Feature Extraction/Vectorization

## Bag Of Words

```python
from sklearn.feature_extraction.text import CountVectorizer 

documents = [
    "I love pizza", 
    "Pizza is the best", 
    "I love pasta", 
    "Pasta is great"
]

vectorizer = CountVectorizer() 

X = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out()) 
print("\n BoW Matrix: \n", X.toarray())
```

## Bigram and Trigram

```python
from sklearn.feature_extraction.text import CountVectorizer 

documents = [
    "I love pizza", 
    "Pizza is the best", 
    "I love pasta", 
    "Pasta is great"
]

vectorizer = CountVectorizer(ngram_range=(2,2)) 

X = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out()) 
print("\n BoW Matrix: \n", X.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer 

documents = [
    "I love pizza", 
    "Pizza is the best", 
    "I love pasta", 
    "Pasta is great"
]

vectorizer = CountVectorizer(ngram_range=(1,3)) 

X = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out()) 
print("\n BoW Matrix: \n", X.toarray())
```

## TF-IDF 

```python
from sklearn.feature_extraction.text import TfidfVectorizer 

documents = [
    "I love pizza", 
    "Pizza is the best", 
    "I love pasta", 
    "Pasta is great"
]

vectorizer = TfidfVectorizer() 

X = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out()) 
print("\n TF-IDF Matrix: \n", X.toarray())
```

## Model Building

```python
X = df["title"]
y = df['emotions']
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split 
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
```

```python
bow_vect = CountVectorizer() 
```

```python
X_train_bow = bow_vect.fit_transform(X_train)
```

```python
X_train_bow
```

```python
X_test_bow = bow_vect.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB 
from sklearn.metrics import accuracy_score
```

```python
nb_model = MultinomialNB() 
```

```python
nb_model.fit(X_train_bow, y_train)
```

```python
y_pred_nb = nb_model.predict(X_test_bow)
```

```python
accuracy_score(y_test, y_pred_nb)
```

```python
tdidf_model = TfidfVectorizer() 

X_train_tfidf = tdidf_model.fit_transform(X_train)
X_test_tfidf = tdidf_model.transform(X_test)
```

```python
nb_model2 = MultinomialNB() 
nb_model2.fit(X_train_tfidf, y_train)
```

```python
tfidf_pred = nb_model2.predict(X_test_tfidf)

accuracy_score(y_test, tfidf_pred)
```

## Logistic Regression with TF-IDF

```python
from sklearn.linear_model import LogisticRegression 

lr_model = LogisticRegression(max_iter=1000) 

lr_model.fit(X_train_tfidf, y_train)
```

```python
y_lr_pred = lr_model.predict(X_test_tfidf)

accuracy_score(y_test, y_lr_pred)
```
